##  Problem Statement
The goal of this project is to analyze unemployment trends over time and identify key patterns and insights.

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

##  Load Dataset

```python
# Load dataset (make sure file is uploaded in Jupyter)
df = pd.read_csv("Unemployment.csv")

df.head()
```

```python
# Check dataset info
df.info()
```

##  Data Cleaning
Removing missing values and preparing the dataset.

```python
# Remove missing values
df = df.dropna()
```

```python
##  Unemployment Trend Over Time
```

```python
df.columns = df.columns.str.strip()
```

```python
# Convert Date column
df["Date"] = pd.to_datetime(df["Date"])

# Plot
plt.figure(figsize=(10,5))
plt.plot(df["Date"], df["Estimated Unemployment Rate (%)"])
plt.title("Unemployment Rate Over Time")
plt.xlabel("Date")
plt.ylabel("Estimated Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

##  Distribution of Unemployment Rate

```python
sns.histplot(df["Estimated Unemployment Rate (%)"], kde=True)
plt.title("Distribution of Unemployment Rate")
plt.show()
```

##  Covid-19 Impact Analysis

```python
# Filter Covid period (2020)
covid_data = df[df["Date"].dt.year == 2020]

plt.figure(figsize=(10,5))
plt.plot(covid_data["Date"], covid_data["Estimated Unemployment Rate (%)"])
plt.title("Unemployment During Covid-19 (2020)")
plt.xticks(rotation=45)
plt.show()
```

##  Insights

- Unemployment rates vary over time.
- A noticeable increase can be observed during the Covid-19 period.
- Data shows clear trends and patterns useful for analysis.

##  Conclusion

The analysis shows that unemployment trends changed significantly over time.  
During the Covid-19 period, unemployment rates increased noticeably, highlighting its economic impact.
